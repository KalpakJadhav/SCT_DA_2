```python
import pandas as pd

# Load the dataset
df=pd.read_csv("C:/INTERN/Task-2/superstore.csv");df
```

```python
print("Missing values per column:\n", df.isnull().sum())
df.fillna({'Postal Code': 0}, inplace=True)
```

```python
initial_count = len(df)
df.drop_duplicates(inplace=True)
print(f"\nRemoved {initial_count - len(df)} duplicate rows.")
```

```python
df['Order.Date'] = pd.to_datetime(df['Order.Date'])
df['Ship.Date'] = pd.to_datetime(df['Ship.Date'])

df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')

print("\nNew Data Types:\n", df.dtypes[['Order.Date', 'Ship.Date', 'Sales', 'Profit']])
```

```python
df.to_csv('Superstore_Cleaned_Data.csv', index=False)
print("\nSuccess! The cleaned file has been saved as 'Superstore_Cleaned_Data.csv'")
```
